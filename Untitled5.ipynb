```python
import re
import random
import torch
import torch.nn as nn
import torch.optim as optim


# ============================================================
# SETTINGS
# ============================================================

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

EMBED_SIZE = 128
HIDDEN_SIZE = 256
LEARNING_RATE = 0.001
EPOCHS = 500


# ============================================================
# DATASET
# English -> Hindi + Marathi
# ============================================================

data = [
    ("hello", "नमस्ते", "नमस्कार"),
    ("good morning", "सुप्रभात", "शुभ प्रभात"),
    ("good night", "शुभ रात्रि", "शुभ रात्री"),
    ("how are you", "आप कैसे हैं", "तुम्ही कसे आहात"),
    ("i am fine", "मैं ठीक हूँ", "मी ठीक आहे"),
    ("what is your name", "आपका नाम क्या है", "तुमचे नाव काय आहे"),
    ("my name is rahul", "मेरा नाम राहुल है", "माझे नाव राहुल आहे"),
    ("where are you going", "आप कहाँ जा रहे हैं", "तुम्ही कुठे जात आहात"),
    ("i am going home", "मैं घर जा रहा हूँ", "मी घरी जात आहे"),
    ("i like tea", "मुझे चाय पसंद है", "मला चहा आवडतो"),
    ("i like coffee", "मुझे कॉफी पसंद है", "मला कॉफी आवडते"),
    ("i am hungry", "मुझे भूख लगी है", "मला भूक लागली आहे"),
    ("i am thirsty", "मुझे प्यास लगी है", "मला तहान लागली आहे"),
    ("please help me", "कृपया मेरी मदद करें", "कृपया मला मदत करा"),
    ("thank you", "धन्यवाद", "धन्यवाद"),
    ("you are welcome", "आपका स्वागत है", "तुमचे स्वागत आहे"),
    ("see you tomorrow", "कल मिलते हैं", "उद्या भेटू"),
    ("what are you doing", "आप क्या कर रहे हैं", "तुम्ही काय करत आहात"),
    ("i am studying", "मैं पढ़ रहा हूँ", "मी अभ्यास करत आहे"),
    ("i am working", "मैं काम कर रहा हूँ", "मी काम करत आहे"),
    ("this is my house", "यह मेरा घर है", "हे माझे घर आहे"),
    ("this is my book", "यह मेरी किताब है", "हे माझे पुस्तक आहे"),
    ("i love india", "मुझे भारत से प्यार है", "मला भारत आवडतो"),
    ("india is beautiful", "भारत सुंदर है", "भारत सुंदर आहे"),
    ("what time is it", "अभी कितने बजे हैं", "आत्ता किती वाजले आहेत"),
    ("i don't know", "मुझे नहीं पता", "मला माहित नाही"),
    ("i understand", "मैं समझता हूँ", "मला समजते"),
    ("i don't understand", "मैं नहीं समझता", "मला समजत नाही"),
    ("come here", "यहाँ आओ", "इकडे या"),
    ("go there", "वहाँ जाओ", "तिकडे जा"),
]


# ============================================================
# TOKENIZER
# ============================================================

def tokenize(sentence):

    sentence = sentence.lower().strip()

    sentence = re.sub(
        r"([?.!,])",
        r" \1",
        sentence
    )

    return sentence.split()


# ============================================================
# VOCABULARY
# ============================================================

class Vocabulary:

    def __init__(self):

        self.word_to_idx = {
            "<PAD>": 0,
            "<SOS>": 1,
            "<EOS>": 2,
            "<UNK>": 3
        }

        self.idx_to_word = {
            0: "<PAD>",
            1: "<SOS>",
            2: "<EOS>",
            3: "<UNK>"
        }

    def add_word(self, word):

        if word not in self.word_to_idx:

            idx = len(self.word_to_idx)

            self.word_to_idx[word] = idx
            self.idx_to_word[idx] = word

    def build(self, sentences):

        for sentence in sentences:

            for word in tokenize(sentence):
                self.add_word(word)

    def encode(self, sentence):

        tokens = tokenize(sentence)

        ids = [
            self.word_to_idx.get(
                word,
                self.word_to_idx["<UNK>"]
            )
            for word in tokens
        ]

        return (
            [self.word_to_idx["<SOS>"]]
            + ids
            + [self.word_to_idx["<EOS>"]]
        )

    def decode(self, ids):

        words = []

        for idx in ids:

            word = self.idx_to_word.get(
                idx,
                "<UNK>"
            )

            if word == "<EOS>":
                break

            if word not in ["<SOS>", "<PAD>"]:
                words.append(word)

        return " ".join(words)

    def __len__(self):
        return len(self.word_to_idx)


# ============================================================
# BUILD ONE COMMON VOCABULARY
# ============================================================

all_sentences = []

for english, hindi, marathi in data:

    all_sentences.append(english)
    all_sentences.append(hindi)
    all_sentences.append(marathi)


vocab = Vocabulary()

vocab.build(all_sentences)

# Add language control tokens
vocab.add_word("<HIN>")
vocab.add_word("<MAR>")


# ============================================================
# ENCODER
# ============================================================

class Encoder(nn.Module):

    def __init__(
        self,
        vocab_size,
        embedding_size,
        hidden_size
    ):

        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_size,
            padding_idx=0
        )

        self.lstm = nn.LSTM(
            embedding_size,
            hidden_size,
            batch_first=True
        )

    def forward(self, source):

        embedded = self.embedding(source)

        _, (hidden, cell) = self.lstm(
            embedded
        )

        return hidden, cell


# ============================================================
# DECODER
# ============================================================

class Decoder(nn.Module):

    def __init__(
        self,
        vocab_size,
        embedding_size,
        hidden_size
    ):

        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_size,
            padding_idx=0
        )

        self.lstm = nn.LSTM(
            embedding_size,
            hidden_size,
            batch_first=True
        )

        self.fc = nn.Linear(
            hidden_size,
            vocab_size
        )

    def forward(
        self,
        input_token,
        hidden,
        cell
    ):

        input_token = input_token.unsqueeze(1)

        embedded = self.embedding(input_token)

        output, (hidden, cell) = self.lstm(
            embedded,
            (hidden, cell)
        )

        prediction = self.fc(
            output.squeeze(1)
        )

        return prediction, hidden, cell


# ============================================================
# SEQ2SEQ
# ============================================================

class Seq2Seq(nn.Module):

    def __init__(
        self,
        encoder,
        decoder,
        device
    ):

        super().__init__()

        self.encoder = encoder
        self.decoder = decoder
        self.device = device

    def forward(
        self,
        source,
        target,
        teacher_forcing_ratio=0.5
    ):

        batch_size = source.size(0)

        target_length = target.size(1)

        vocab_size = (
            self.decoder.fc.out_features
        )

        outputs = torch.zeros(
            batch_size,
            target_length,
            vocab_size
        ).to(self.device)

        hidden, cell = self.encoder(source)

        input_token = target[:, 0]

        for t in range(1, target_length):

            output, hidden, cell = self.decoder(
                input_token,
                hidden,
                cell
            )

            outputs[:, t] = output

            best_prediction = output.argmax(1)

            if random.random() < teacher_forcing_ratio:

                input_token = target[:, t]

            else:

                input_token = best_prediction

        return outputs


# ============================================================
# CREATE MODEL
# ============================================================

encoder = Encoder(
    len(vocab),
    EMBED_SIZE,
    HIDDEN_SIZE
)

decoder = Decoder(
    len(vocab),
    EMBED_SIZE,
    HIDDEN_SIZE
)

model = Seq2Seq(
    encoder,
    decoder,
    DEVICE
).to(DEVICE)


# ============================================================
# PREPARE TRAINING DATA
# ============================================================

training_data = []

for english, hindi, marathi in data:

    # English input
    source = vocab.encode(english)

    # Create Hindi target
    hindi_target = (
        ["<HIN>"]
        + tokenize(hindi)
    )

    hindi_ids = [
        vocab.word_to_idx["<SOS>"]
    ]

    hindi_ids += [
        vocab.word_to_idx.get(
            word,
            vocab.word_to_idx["<UNK>"]
        )
        for word in hindi_target
    ]

    hindi_ids.append(
        vocab.word_to_idx["<EOS>"]
    )

    # Create Marathi target
    marathi_target = (
        ["<MAR>"]
        + tokenize(marathi)
    )

    marathi_ids = [
        vocab.word_to_idx["<SOS>"]
    ]

    marathi_ids += [
        vocab.word_to_idx.get(
            word,
            vocab.word_to_idx["<UNK>"]
        )
        for word in marathi_target
    ]

    marathi_ids.append(
        vocab.word_to_idx["<EOS>"]
    )

    training_data.append(
        (
            torch.tensor(source),
            torch.tensor(hindi_ids)
        )
    )

    training_data.append(
        (
            torch.tensor(source),
            torch.tensor(marathi_ids)
        )
    )


# ============================================================
# LOSS + OPTIMIZER
# ============================================================

optimizer = optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

criterion = nn.CrossEntropyLoss(
    ignore_index=vocab.word_to_idx["<PAD>"]
)


# ============================================================
# TRAIN
# ============================================================

print("\n================================")
print("Training English -> Hindi/Marathi")
print("================================")

model.train()

for epoch in range(EPOCHS):

    total_loss = 0

    random.shuffle(training_data)

    for source, target in training_data:

        source = source.unsqueeze(0).to(DEVICE)

        target = target.unsqueeze(0).to(DEVICE)

        optimizer.zero_grad()

        output = model(
            source,
            target,
            teacher_forcing_ratio=0.5
        )

        output = output[:, 1:, :]

        target = target[:, 1:]

        output = output.reshape(
            -1,
            output.shape[-1]
        )

        target = target.reshape(-1)

        loss = criterion(
            output,
            target
        )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            1
        )

        optimizer.step()

        total_loss += loss.item()

    if (epoch + 1) % 50 == 0:

        avg_loss = (
            total_loss / len(training_data)
        )

        print(
            f"Epoch {epoch + 1}/{EPOCHS} "
            f"Loss: {avg_loss:.4f}"
        )


# ============================================================
# TRANSLATION FUNCTION
# ============================================================

def translate(sentence, language):

    model.eval()

    # English input
    source_ids = vocab.encode(sentence)

    source = torch.tensor(
        source_ids,
        dtype=torch.long
    ).unsqueeze(0).to(DEVICE)

    # Encode
    with torch.no_grad():

        hidden, cell = model.encoder(
            source
        )

    # Select language
    if language == "hindi":

        language_token = "<HIN>"

    else:

        language_token = "<MAR>"

    # First decoder token
    input_token = torch.tensor(
        [
            vocab.word_to_idx["<SOS>"]
        ],
        dtype=torch.long
    ).to(DEVICE)

    translated = []

    # First generate language token
    with torch.no_grad():

        output, hidden, cell = model.decoder(
            input_token,
            hidden,
            cell
        )

    # Force decoder to start with selected language
    input_token = torch.tensor(
        [
            vocab.word_to_idx[language_token]
        ],
        dtype=torch.long
    ).to(DEVICE)

    max_length = 30

    with torch.no_grad():

        for _ in range(max_length):

            output, hidden, cell = model.decoder(
                input_token,
                hidden,
                cell
            )

            prediction = output.argmax(1)

            token_id = prediction.item()

            if token_id == vocab.word_to_idx["<EOS>"]:

                break

            word = vocab.idx_to_word.get(
                token_id,
                "<UNK>"
            )

            if word not in [
                "<SOS>",
                "<PAD>",
                "<HIN>",
                "<MAR>"
            ]:

                translated.append(word)

            input_token = prediction

    return " ".join(translated)


# ============================================================
# USER INTERFACE
# ============================================================

print("\n================================")
print(" ENGLISH TRANSLATION MODEL")
print("================================")

while True:

    print("\nEnter 'quit' to exit.")

    sentence = input(
        "\nEnter English sentence: "
    ).strip()

    if sentence.lower() == "quit":
        print("Goodbye!")
        break

    if not sentence:
        print("Please enter a sentence.")
        continue

    # Ask language
    print("\nChoose translation language:")
    print("1. Hindi")
    print("2. Marathi")

    choice = input(
        "Enter choice (1/2): "
    ).strip()

    if choice == "1":

        language = "hindi"

    elif choice == "2":

        language = "marathi"

    else:

        print("Invalid choice. Please select 1 or 2.")
        continue

    # Translate
    result = translate(
        sentence,
        language
    )

    print(
        f"\n{language.capitalize()} translation:"
    )

    print(result)
```
