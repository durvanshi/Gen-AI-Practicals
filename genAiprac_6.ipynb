```python
!pip install -q transformers sentencepiece

import transformers
import textwrap
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

print(f"Transformers version: {transformers.__version__}")

# Load pretrained BART model
model_name = "facebook/bart-large-cnn"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# News article
article = """
Artificial intelligence is rapidly transforming industries across the world.
Companies are increasingly using AI to automate repetitive tasks, analyze
large amounts of data, and improve customer experiences. Experts believe that
AI will create new opportunities in areas such as healthcare, education,
finance, and transportation. However, the rapid adoption of AI has also raised
concerns about privacy, job displacement, misinformation, and the ethical use
of automated systems.

Governments and technology companies are working on regulations and
guidelines
to ensure that AI is developed and used responsibly. Researchers are also
developing methods to make AI systems more transparent, secure, and
reliable.
As the technology continues to evolve, organizations are expected to invest
more heavily in AI while focusing on responsible and human-centered
innovation.
"""

# Tokenize the article
inputs = tokenizer(
    article,
    return_tensors="pt",
    max_length=1024,
    truncation=True
)

# Generate summary
summary_ids = model.generate(
    inputs["input_ids"],
    max_length=100,
    min_length=30,
    num_beams=4,
    early_stopping=True
)

# Convert tokens back to text
summary_text = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print("\nOriginal Article:\n")
print(article)

print("\nGenerated Summary:\n")
print(textwrap.fill(summary_text, width=80))
```
